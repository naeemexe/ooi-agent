```python
import matplotlib.pyplot as plt
import numpy as np
from ooi_tools import DATASETS

ds = DATASETS["CE02SHSM"]
var = "fluorometric_chlorophyll_a"

years = sorted(set(ds.time.dt.year.values))
n = len(years)
fig, axes = plt.subplots(n, 1, figsize=(12, 3 * n), sharex=False)

if n == 1:
    axes = [axes]

for ax, yr in zip(axes, years):
    subset = ds.sel(time=str(yr))
    if subset.time.size > 0:
        ax.plot(subset.time.values, subset[var].values, linewidth=0.5)
    ax.set_ylabel("Chl-a (mg/m³)")
    ax.set_title(str(yr))
    ax.grid(alpha=0.3)

plt.suptitle("Chlorophyll-a — Oregon Shelf (CE02SHSM) — By Year", y=1.01, fontsize=14)
plt.tight_layout()
plt.show()
```

### What to look for
- **Spring bloom**: Typically March–May on the Oregon shelf, triggered by increasing light + residual nutrients before upwelling fully starts.
- **Summer blooms**: Can occur during relaxation events (wind relaxes → stratification → surface bloom).
- **Fall/winter**: Low chlorophyll due to mixing and low light.
- **Bloom timing**: Look at the peak dates — that's when phytoplankton biomass was highest. Compare across years to see if timing shifts.
